## Observações: 

### variáveis numéricas

- A variável 'declared_monthly_revenue' possui preenchimento completo (842 registros), mas sua distribuição é extremamente concentrada em zero: primeiro quartil, mediana e terceiro quartil são iguais a 0. A média (73.377) é inflada por poucos valores extremos, chegando a um máximo de 50 milhões. A elevada assimetria (28) e curtose (796) confirmam forte influência de outliers, tornando a variável pouco informativa.

- A variável 'declared_product_catalog_size' apresenta uma distribuição plausível (mediana = 100; média = 233; assimetria = 2,67), porém conta com apenas 69 registros preenchidos entre os 842 negócios, comprometendo sua representatividade e utilidade analítica.

### variáveis temporais

- A captação de leads ('first_contact_date') ocorreu entre 14/06/2017 e 31/05/2018, enquanto os fechamentos ('won_date') foram registrados entre 05/12/2017 e 14/11/2018. Como os fechamentos continuaram por cerca de cinco meses após o encerramento da captação, os leads mais recentes tiveram menos tempo para converter do que os mais antigos. Essa diferença na janela de observação deve ser considerada nas análises posteriores. 

- A captação acelerou significativamente em 2018, com mais de mil leads por mês entre janeiro e maio, contra algumas centenas em 2017. Já os fechamentos cresceram até abril de 2018 (207 negócios) e passaram a cair nos meses seguintes. Essa redução reflete principalmente o menor tempo de maturação dos leads mais recentes, e não necessariamente uma piora no desempenho comercial.

- A análise por dia da semana mostra que a captação ocorre continuamente, inclusive aos fins de semana (cerca de 530 leads aos sábados e domingos). Em contrapartida, os fechamentos concentram-se em dias úteis, com apenas três registros aos sábados e nenhum aos domingos, indicando que a geração de leads é automatizada, enquanto a conversão depende da atuação da equipe comercial.

### variáveis categoricas

- A variável 'origin' (canal de captação) é liderada por 'organic_search' (28,7%), 'paid_search' (19,8%) e 'social' (16,9%). Contudo, 13,7% dos registros estão classificados como 'unknown' e 0,8% são nulos, totalizando cerca de 14,5% de leads sem origem identificada. Além disso, as categorias 'other' e 'other_publicities' (2,7% combinadas) possuem baixa especificidade. Esses valores devem ser tratados antes de comparações de conversão entre canais.

- A variável 'business_type' apresenta boa qualidade de dados. 'reseller' (69,7%) e 'manufacturer' (28,7%) concentram praticamente todos os registros, com apenas 1,2% de valores nulos e 0,4% classificados como 'other'. Trata-se da variável categórica mais consistente do banco.

- Na variável 'lead_type', predominam perfis online de diferentes portes ('online_medium', 'online_big', 'online_small', 'online_beginner' e 'online_top'), que juntos representam cerca de 72% dos registros. Os demais casos dividem-se entre 'industry' (14,6%) e 'offline' (12,3%).

- A variável 'business_segment' apresenta uma distribuição de cauda longa. Embora os segmentos mais frequentes, liderados por 'home_decor' e 'health_beauty', concentrem parcela relevante dos negócios, há um grande número de categorias com poucos registros, o que pode dificultar análises segmentadas.

- Por fim, 'lead_behaviour_profile' é um campo multivalorado, contendo combinações de perfis (por exemplo, 'cat, wolf'), além de apresentar 21% de valores ausentes. Como não está relacionado às perguntas analíticas deste projeto, o campo será mantido apenas para possíveis análises futuras.

### variáveis booleanas

- As variáveis 'has_company' e 'has_gtin' apresentam aproximadamente 92% de valores ausentes (779 e 778 registros, respectivamente), o que compromete sua utilidade analítica. Entre os poucos casos preenchidos, predominam valores verdadeiros. Esse padrão reforça a decisão de excluir ambas das análises.
